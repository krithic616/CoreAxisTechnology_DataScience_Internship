```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
df = pd.read_csv('../data/Advertising_cleaned.csv')
df.head()
```

```python
df.shape
df.isnull().sum()
df.duplicated().sum()
```

## Exploratory Data Analysis
TV has the strongest correlation with Sales, followed by Radio. Newspaper has a weaker positive relationship.


```python
df.corr(numeric_only=True)['Sales'].sort_values(ascending=False)
```

```python
X=df[['TV','Radio','Newspaper']]
y=df['Sales']
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.20,random_state=42)
```

```python
lr=LinearRegression().fit(X_train,y_train)
rf=RandomForestRegressor(n_estimators=300,random_state=42).fit(X_train,y_train)
for name,m in [('Linear Regression',lr),('Random Forest',rf)]:
 p=m.predict(X_test)
 print(name, 'MAE=',mean_absolute_error(y_test,p),'RMSE=',np.sqrt(mean_squared_error(y_test,p)),'R2=',r2_score(y_test,p))
```

```python
pd.DataFrame({'Feature':X.columns,'Importance':rf.feature_importances_}).sort_values('Importance',ascending=False)
```

## Conclusion
Random Forest achieved the best test performance. TV is the strongest individual advertising predictor, followed by Radio. Newspaper contributes comparatively less predictive information.
