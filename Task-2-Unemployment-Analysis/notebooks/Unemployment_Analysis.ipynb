## Objective
Analyze unemployment trends, regional differences, rural/urban patterns, and the early COVID-19 period using the exact CoreAxis dataset.

```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

DATA_PATH = Path('../data/Unemployment in India.csv')
df = pd.read_csv(DATA_PATH)
df.columns = df.columns.str.strip()
df = df.dropna(how='all').drop_duplicates().copy()
df['Date'] = pd.to_datetime(df['Date'], dayfirst=True, errors='coerce')
numeric_cols = ['Estimated Unemployment Rate (%)','Estimated Employed','Estimated Labour Participation Rate (%)']
for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')
df.head()
```

```python
monthly = df.groupby('Date').agg(unemployment_rate=('Estimated Unemployment Rate (%)','mean'), employed=('Estimated Employed','mean'), labour_participation=('Estimated Labour Participation Rate (%)','mean')).reset_index()
monthly
```

```python
plt.figure(figsize=(10,5))
plt.plot(monthly['Date'], monthly['unemployment_rate'], marker='o')
plt.axvline(pd.Timestamp('2020-03-31'), linestyle='--')
plt.title('Monthly Average Unemployment Rate')
plt.xlabel('Date'); plt.ylabel('Average unemployment rate (%)')
plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

```python
area = df.groupby('Area')['Estimated Unemployment Rate (%)'].mean().sort_values()
plt.figure(figsize=(8,5)); plt.bar(area.index, area.values)
plt.title('Average Unemployment Rate: Rural vs Urban')
plt.ylabel('Average unemployment rate (%)'); plt.tight_layout(); plt.show()
```

```python
region = df.groupby('Region')['Estimated Unemployment Rate (%)'].mean().sort_values()
plt.figure(figsize=(10,8)); plt.barh(region.index, region.values)
plt.title('Average Unemployment Rate by Region')
plt.xlabel('Average unemployment rate (%)'); plt.tight_layout(); plt.show()
```

```python
covid = monthly[monthly['Date'] >= pd.Timestamp('2020-01-31')]
plt.figure(figsize=(9,5)); plt.plot(covid['Date'], covid['unemployment_rate'], marker='o')
plt.axvline(pd.Timestamp('2020-03-31'), linestyle='--')
plt.title('Unemployment Rate During Early COVID-19 Period')
plt.ylabel('Average unemployment rate (%)'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```
